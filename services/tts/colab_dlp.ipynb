```python
# 1. 安裝必要工具
!pip install yt-dlp demucs
# 2. 設定 YouTube 網址與檔名
URL = "https://www.youtube.com/watch?v=TsV8hG1SaEw"
NAME = "nexus_vocal_source"
# 3. 強力抓取音訊
!yt-dlp -x --audio-format wav --audio-quality 0 \
--extractor-args "youtube:player_client=android,web" \
-o "{NAME}.wav" "{URL}"

# 4. 啟動靈魂分離 (Demucs)
!demucs --two-stems=vocals -n hdemucs_mmi "{NAME}.wav"

# 5. 打包並下載回本地
import os
from google.colab import files

vocal_path = f"separated/hdemucs_mmi/{NAME}/vocals.wav"
if os.path.exists(vocal_path):
  os.rename(vocal_path, f"{NAME}_pure_vocals.wav")
  files.download(f"{NAME}_pure_vocals.wav")
  print("✅ 成功！純淨人聲已準備好下載到你的電腦囉~ (◍•ᴗ•◍)ﾉ♡")
else:
  print("❌ 分離失敗，請檢查 Log 訊息。")
```
